```python
import pandas as pd
import mysql.connector
from pathlib import Path

# 1. Cleaned CSV file path
file_path = Path(
    "../data/ecommerce_project_cleaned_dataset/order_items.csv"
)

# 2. Connect to MySQL
conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="Pujitha@1234",
    database="ecommerce_analysis"
)

cursor = conn.cursor()

# 3. Prepare the insert query
sql = """
INSERT INTO order_items (
    order_item_id, order_id, product_id, quantity,
    unit_price, discount_amount, line_total, delivery_date
)
VALUES (%s, %s, %s, %s, %s, %s, %s, %s)
"""

# 4. Read and insert 5,000 rows at a time
total_loaded = 0

for chunk in pd.read_csv(
    file_path,
    chunksize=5000,
    na_values=["NULL", ""]
):
    chunk["delivery_date"] = pd.to_datetime(
        chunk["delivery_date"], errors="coerce"
    )

    chunk["delivery_date"] = chunk["delivery_date"].apply(
        lambda x: x.to_pydatetime() if pd.notna(x) else None
    )

    chunk = chunk.astype(object).where(
        pd.notna(chunk), None
    )

    rows = list(chunk.itertuples(index=False, name=None))

    cursor.executemany(sql, rows)
    conn.commit()

    total_loaded += len(rows)
    print("Rows loaded:", total_loaded)

print("Import completed!")

cursor.close()
conn.close()
```
